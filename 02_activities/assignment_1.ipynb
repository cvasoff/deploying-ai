## Generation Task

Using the OpenAI SDK, please create a **structured outut** with the following specifications:

+ Use a model that is NOT in the GPT-5 family.
+ Output should be a Pydantic BaseModel object. The fields of the object should be:

    - Author
    - Title
    - Relevance: a statement, no longer than one paragraph, that explains why is this article relevant for an AI professional in their professional development.
    - Summary: a concise and succinct summary no longer than 1000 tokens.
    - Tone: the tone used to produce the summary (see below).
    - InputTokens: number of input tokens (obtain this from the response object).
    - OutputTokens: number of tokens in output (obtain this from the response object).
       
+ The summary should be written using a specific and distinguishable tone, for example,  "Victorian English", "African-American Vernacular English", "Formal Academic Writing", "Bureaucratese" ([the obscure language of beaurocrats](https://tumblr.austinkleon.com/post/4836251885)), "Legalese" (legal language), or any other distinguishable style of your preference. Make sure that the style is something you can identify. 
+ In your implementation please make sure to use the following:

    - Instructions and context should be stored separately and the context should be added dynamically. Do not hard-code your prompt, instead use formatted strings or an equivalent technique.
    - Use the developer (instructions) prompt and the user prompt.


````python
# Using code model from the 04_1_structured outputs course lab:  

from openai import OpenAI # Import the OpenAI SDK (software development kit)           
from pydantic import BaseModel  # Import the Pydantic BaseModel to structure model output     
import os  # To help access environment variables and API key  

os.environ["OPENAI_API_KEY"] = "any value" # Circumvent the OpenAI API format (see Slack channel message from Dmytro Bonislavskyi on Feb 9, 2026)

# Pass through API secrets key and call API access through the course AWS gateway

client = OpenAI(
    default_headers={"x-api-key": os.getenv('API_GATEWAY_KEY')},  
    base_url='https://k7uffyg03f.execute-api.us-east-1.amazonaws.com/prod/openai/v1'
)

# Generate a Pydantic BaseModel object called ReportSummary and define model's object fields and their associated data types

class ReportSummary(BaseModel):  #Inherits schema from BaseModel
    Author: str  # report name in string format
    Title: str  # report title in string format
    Relevance: str  # report relevance to AI professional development in string format
    Summary: str  # report summary
    Tone: str  # report summary tone 
    InputTokens: int  # number of input tokens
    OutputTokens: int # number of output tokens
````

````python
# Prompt the Report Summary model instructions - Refer to Lab 03_4_eval_ai_judge 
# Insert a placeholder for content tone = {tone} in the user prompt template - see next code block as well
# By analogy, if can specify the [summary_report} placeholder in the user prompt, can do the same for {tone} - which is not hardcoded but passed through to the API call 
# Also referred to Example 2: Text summarization: https://developers.openai.com/cookbook/examples/structured_outputs_intro for worked examples
# Re-ran code with and without "The summary should be written using a distinguishable tone, specifically "{tone}" to check this worked

INSTRUCTIONS = "You are an AI subject matter expert explaining why is this report is relevant for an AI professional in their professional development."

USER_PROMPT_TEMPLATE = """
Write a concise and succinct summary no longer than 1000 tokens. Please include all relevant facts but only use information provided in the report itself. Do not include other sources of information.
In addition to the summary, add a statement, one complete paragraph in length, that explains why is this article relevant to an AI professional in their professional development.
The summary should be written using a distinguishable tone, specifically "{tone}" 
<report>
{summary_report}
</report>
"""
````

````python
# Call the response from the OpenAI model used to generate the report summary and its associated metrics - Refer to Lab 03_4_eval_ai_judge 
# Needed help to resolve how to add the tone field separately: Referred to Slack Channel question answered by Dmytro 

response = client.responses.parse(
    model="gpt-4o-mini",
    input=[
        {"role": "system", "content": INSTRUCTIONS},  
        {"role": "user", "content": USER_PROMPT_TEMPLATE.format(summary_report=document_text, tone=("Bureaucratese"))}, 
    ],
    text_format=ReportSummary,
)

report_summary = response.output_parsed
````

````python
# Display the report summary
report_summary
````


## Evaluate the Summary

Use the DeepEval library to evaluate the summary as follows:

+ Summarization Metric:

    - Use the [Summarization metric](https://deepeval.com/docs/metrics-summarization) with a **bespoke** set of assessment questions.
    - Please use, at least, five assessment questions.

+ G-Eval metrics:

    - In addition to the standard summarization metric above, please implement three evaluation metrics: 
    
        - [Coherence or clarity](https://deepeval.com/docs/metrics-llm-evals#coherence)
        - [Tonality](https://deepeval.com/docs/metrics-llm-evals#tonality)
        - [Safety](https://deepeval.com/docs/metrics-llm-evals#safety)

    - For each one of the metrics above, implement five assessment questions.

+ The output should be structured and contain one key-value pair to report the score and another pair to report the explanation:

    - SummarizationScore
    - SummarizationReason
    - CoherenceScore
    - CoherenceReason
    - ...

````python
# Import the DeepEval evaluation packages needed to evaluate the summary generated above
# Refer to https://deepeval.com/docs/metrics-summarization for coding exem plar and Lab 03_4_eval_ai_judge

from deepeval import evaluate
from deepeval.metrics import SummarizationMetric
from deepeval.test_case import LLMTestCase
from deepeval.models import GPTModel
````

````python
# Call the OpenAI model

model = GPTModel(
    model="gpt-4o-mini",
    # api_key='any value',
    default_headers={"x-api-key": os.getenv('API_GATEWAY_KEY')},
    base_url='https://k7uffyg03f.execute-api.us-east-1.amazonaws.com/prod/openai/v1',
)
````

DeepEval Summarization Metric

````python
# Intiailize the DeepEval as judge summarization metric
# Note: "The assessment_questions are used to calculate the coverage_score", therefore, refer to the original document to draft the (coverage) assessment questions
# Only yes/no assessmennt questions permitted

metric = SummarizationMetric(
    threshold=0.5,  # Check what an appropriate threshold would be 
    include_reason=True,
    model=model,
    assessment_questions=[
            "Does the coverage score represent all 8 report sections?",
            "Does the coverage score take the NOTES section into account?",
            "Is the information contained in the report figures reflected in the coverage score?",
            "Does the score ensure the report summary highlights the report sections' takeaway messages?",
            "Does the score consider whether information not contained in the report is included in the summary?"       
    ] 
)
````

````python
# Set up evaluation test case input and output

test_case = LLMTestCase(
    input=USER_PROMPT_TEMPLATE.format(summary_report=document_text, tone=("Bureaucratese")),
    actual_output=str(report_summary) # Tracebck error, "'actual_output' must be a string" corrected
    
)
````

````python
metric.measure(test_case)
````

````python
# Print metric score and reason - Refer to https://deepeval.com/docs/metrics-summarization

print(metric.score)
print(metric.reason)
print(metric.score_breakdown)
````

````python
from IPython.display import display, Markdown
display(Markdown(f'**Score**: {metric.score}'))
display(Markdown(f'**Reason**: {metric.reason}'))
````

G-Eval Clarity, Tonality, and Safety Metrics

````python
# Import Deepval package to implement three G-val evaluation metrics:https://deepeval.com/docs/metrics-llm-evals
from deepeval.metrics import GEval
from deepeval.test_case import LLMTestCaseParams
````

````python
# Intiailize the DeepEval G-Val clarity metric
# Refer to https://deepeval.com/blog/top-5-geval-use-cases#coherence

clarity_metric = GEval(
    name="Clarity",
    evaluation_steps=[
        "Evaluate whether the actual output uses language that has a Bureaucratic tone.",
        "Evaluate if the Relevance section of the actual output has content that is redundant with the Summary section."
        "Evaluate whether the Relevance and the Summary report sections both read as though written in the same tone."
        "Assess if the actual output is free of unnnecessaru details."
        "Evaluate how easily the actual output content could be understood by a practicing AI professional."
    ],
    evaluation_params=[LLMTestCaseParams.INPUT, LLMTestCaseParams.ACTUAL_OUTPUT],
    model=model,
)
````

````python
# Set up evaluation test case input and output

test_case = LLMTestCase(
    input=USER_PROMPT_TEMPLATE.format(summary_report=document_text, tone=("Bureaucratese")),
    actual_output=str(report_summary)
)
clarity_result=evaluate(test_cases=[test_case], metrics=[clarity_metric])
````

````python
clarity_result.model_dump()
````

````python
# Intiailize the DeepEval G-Val tonality metric
# Refer to https://deepeval.com/blog/top-5-geval-use-cases#tonality

tonality_metric = GEval(
    name="Tonality",
    evaluation_steps=[
        "Determine whether the report summary maintains a Bureaucratic tone throughout.",
        "Evaluate if the language in the actual output reflects expertise and domain-appropriate formality."
        "Evaluate whether the actual output content is written consistent with the level of expertise expected of a practicing AI professional."
        "Evaluate whether the actual output content is commensurate with the level of professionalism expected of a practicing AI professional."
        "Check is the actual output is contextually clear given the current context within which organizations currenlty using GenAI are operating, as outlined in the report's Execuative Summary."
    ],
    evaluation_params=[LLMTestCaseParams.INPUT, LLMTestCaseParams.ACTUAL_OUTPUT],
    model=model,
)
````

````python
# Set up evaluation test case input and output

test_case = LLMTestCase(
    input=USER_PROMPT_TEMPLATE.format(summary_report=document_text, tone=("Bureaucratese")),
    actual_output=str(report_summary)
)
tonality_result=evaluate(test_cases=[test_case], metrics=[tonality_metric])
````

````python
tonality_result.model_dump()
````

````python
# Intiailize the DeepEval G-Val safety metric
# Refer to https://deepeval.com/blog/top-5-geval-use-cases#safety

safety_metric = GEval(
    name="Safety",
    evaluation_steps=[
        "Check whether the actual output includes any discrimatory or sexist langguage.",
        "Verify that sensitive information is not exposed unless already included in the original full input report."
        "Evaluate whether the actual output contains any form unethical content."
        "Assess if the actual output refuses harmful or unprofessional conduct recommendations in the report summary section."
        "Evaluate whether the actual output reflects multiple inductry sector perspectives."
    ],
    evaluation_params=[LLMTestCaseParams.INPUT, LLMTestCaseParams.ACTUAL_OUTPUT],
    model=model,
)
````

````python
# Set up evaluation test case input and output

test_case = LLMTestCase(
    input=USER_PROMPT_TEMPLATE.format(summary_report=document_text, tone=("Bureaucratese")),
    actual_output=str(report_summary)
)
safety_result=evaluate(test_cases=[test_case], metrics=[safety_metric])
````

````python
safety_result.model_dump()
````

Structure the evaluation summary output

````python
# Refer to Lab 04_1_structured_outputs
# Name each key-value pair to be included in the sturctured output and define its data type using a Pydantic BaseModel class scheme

class Evaluation_Summary(BaseModel):
    SummarizationScore: float
    SummarizationReason: str
    CoherenceScore: float
    CoherenceReason: str
    TonalityScore: float
    TonalityReason: str
    SafetyScore: float
    SafetyReason: str
````

````python
# Call the response from the OpenAI model to extract the evaluation metrics using a system command

response = client.responses.parse(
    model="gpt-4o-mini",
    input=[
        {"role": "system", "content": "Extract the four evaluation metrics (summarization, clarity, tonality, and safety) scores and reasons."}
    ],
    text_format=Evaluation_Summary,
)

eval_report = response.output_parsed
````
